```python
import numpy as np
from scipy.spatial import KDTree

dim = 25
```

```python
def load_glove_model(path, dim):
    f = open(path, 'r')
    xs = []
    ys = []
    x = 0
    for line in f:
        x += 1
        splitLine = line.split()
        word = splitLine[0]
        embedding = [float(val) for val in splitLine[1:]]
        if (len(embedding) != dim):
            if (len(embedding == 24)):
                embedding.prepend('<bork>')
            print("FUCK", x, line, len(embedding))
        xs.append(word)
        ys.append(embedding)
    return (xs, np.array(ys))

def query_similar(vec, words, tree):
    return words[tree.query(vec)[0][0]]
```

```python
print("Loading GloVe vectors...")
words, vecs = load_glove_model('/data/pretrained/glove/glove.twitter.27B.' + dim + 'd.txt', dim)
print("Done")
```
